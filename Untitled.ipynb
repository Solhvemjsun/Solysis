```python
import matplotlib.pyplot as plt
import numpy as np

initial_anses = np.linspace(1,100,100)
max_iterations = 7
b_values = np.linspace(1, 100, 100)

def iterate_log(ans, b, iterations):
    result = [ans]
    for _ in range(iterations):
        ans = np.log(b) / np.log(ans)
        result.append(ans)
    return result

plt.figure(figsize=(20,16))
for b in b_values:
    for initial_ans in initial_anses:
        iterations = iterate_log(initial_ans, b, max_iterations)
        plt.plot(range(max_iterations + 1), iterations, label=f'b = {b}')

plt.xlabel('Iteration')
plt.ylabel('ans')
plt.title('Convergence of ans = log(ans, b)')
plt.legend()
plt.grid(True)
plt.show()
```
